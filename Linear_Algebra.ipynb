```python
from numpy import linalg as la

print('L2 norm  ||a||_2 :', la.norm(a))         # sqrt(sum of squares)
print('L1 norm  ||a||_1 :', la.norm(a, 1))      # sum of absolute values
print('Linf norm        :', la.norm(a, np.inf)) # max absolute value

# Cosine similarity: the angle between vectors, ignoring magnitude
def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

print('cosine(a, b)     :', round(cosine(a, b), 3))
```

```python
import numpy as np
from numpy import linalg as la

# Cosine similarity function (re-defined for clarity in this cell)
def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),    # pair 1
    (np.array([1, 2, 3]), np.array([3, 2, 1])),    # pair 2
    (np.array([2, 0, 0]), np.array([0, 5, 0])),    # pair 3
]

max_similarity = -1
most_similar_pair_index = -1

for i, (u, v) in enumerate(pairs, start=1):
    print(f'\nPair {i}:')
    # 1. L1 and L2 norm of u
    print(f'  L1 norm of u: {la.norm(u, 1)}')
    print(f'  L2 norm of u: {la.norm(u)}')

    # 2. cosine similarity of u and v
    sim = round(cosine(u, v), 3)
    print(f'  Cosine similarity of u and v: {sim}')

    if sim > max_similarity:
        max_similarity = sim
        most_similar_pair_index = i

# 3. Most similar pair = ?   (write your answer here)
print(f'\nMost similar pair = Pair {most_similar_pair_index} (Cosine Similarity: {max_similarity})')
```

```python
import numpy as np

pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),    # pair 1
    (np.array([1, 2, 3]), np.array([3, 2, 1])),    # pair 2
    (np.array([2, 0, 0]), np.array([0, 5, 0])),    # pair 3
]

for i, (u, v) in enumerate(pairs, start=1):
    # 1. L1 and L2 norm of u
    l1 = np.linalg.norm(u, 1)
    l2 = np.linalg.norm(u, 2)

    # 2. Cosine similarity of u and v
    cosine = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

    print(f"Pair {i}:")
    print("L1 Norm =", l1)
    print("L2 Norm =", l2)
    print("Cosine Similarity =", cosine)
    print()

# 3. Most similar pair
print("Most similar pair = Pair 1")
```

```python
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])

print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))

```

```python
I = np.eye(2)                       # identity: 1s on the diagonal
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))

# Symmetric: A equals its own transpose
print('A symmetric?', np.allclose(A, A.T))

# Orthogonal: Q.T @ Q == I (a pure rotation/reflection)
theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))

```

```python
import numpy as np

M = np.array([[4., 2.],
              [2., 3.]])
P = np.array([[0., -1.],
              [1.,  0.]])   # a 90-degree rotation

# 1. Inverse and trace of M
inv_M = np.linalg.inv(M)
trace_M = np.trace(M)

print("Inverse of M:")
print(inv_M)

print("\nTrace of M:")
print(trace_M)

# 2. M @ inv(M) == I ?
identity = M @ inv_M
print("\nM @ inv(M):")
print(identity)

print("\nIs M @ inv(M) equal to Identity?")
print(np.allclose(identity, np.eye(2)))

# 3. Is M symmetric? Is P orthogonal?

# Check if M is symmetric
is_symmetric = np.array_equal(M, M.T)

# Check if P is orthogonal
is_orthogonal = np.allclose(P.T @ P, np.eye(2))

print("\nIs M symmetric?")
print(is_symmetric)

print("\nIs P orthogonal?")
print(is_orthogonal)
```

```python
# A unit square defined by its 4 corners (each column is a point)
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)

# Scaling matrix: stretch x by 1.5, y by 0.5
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])

# Rotation matrix: rotate by 30 degrees
t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

scaled  = S @ square      # apply scaling
rotated = R @ square      # apply rotation
print('Rotated corners:\n', rotated)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def close_loop(pts):
    # repeat the first point at the end so the polygon closes
    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

tri = np.array([[0, 1, 0.5],
                [0, 0, 1.0]])   # 3 corners of a triangle

# 1. Scaling matrix (x*2, y*0.5)
S = np.array([[2, 0],
              [0, 0.5]])

# 2. Rotation matrix (45 degrees)
theta = np.radians(45)

R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])

# 3. Apply both transformations
scaled = S @ tri
transformed = R @ scaled

# Close the triangles by repeating the first point
tri_closed = np.hstack((tri, tri[:, [0]]))
trans_closed = np.hstack((transformed, transformed[:, [0]]))

# Plot
plt.figure(figsize=(6,6))
plt.plot(tri_closed[0], tri_closed[1], 'bo-', label='Original')
plt.plot(trans_closed[0], trans_closed[1], 'ro-', label='Transformed')

plt.axis('equal')
plt.grid(True)
plt.legend()
plt.title("Original vs Transformed Triangle")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python

A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)     # vals = eigenvalues, vecs columns = eigenvectors
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
for i in range(len(vals)):
    v = vecs[:, i]                 # i-th eigenvector
    lhs = A @ v                    # A v
    rhs = vals[i] * v              # lambda v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
import numpy as np
import scipy.linalg as la

C = np.array([[4., 1.],
              [2., 3.]])

# 1. Eigenvalues and eigenvectors
vals, vecs = la.eig(C)

# 2. Print eigenvalues
print("Eigenvalues:", vals)
print("Eigenvectors:\n", vecs)

# 3. Verify Cv = λv for the first eigenvector
v = vecs[:, 0]       # First eigenvector
lam = vals[0]        # First eigenvalue

print("\nC @ v =")
print(C @ v)

print("\nlambda * v =")
print(lam * v)

print("\nVerification:", np.allclose(C @ v, lam * v))
```

```python
import numpy as np
import scipy.linalg as la

A = np.array([[ 2.,  1., -1.],
              [-3., -1.,  2.],
              [-2.,  1.,  2.]])
b = np.array([8., -11., -3.])

x = la.solve(A, b)
print('Solution x:', x)                    # [2, 3, -1]
print('Rank of A :', np.linalg.matrix_rank(A))     # 3 -> full rank
print('Full rank -> a unique solution exists')
```

```python
import numpy as np
from scipy.spatial.distance import cosine

king  = np.array([0.8, 0.6, 0.1, 0.2])
queen = np.array([0.7, 0.7, 0.1, 0.3])
apple = np.array([0.1, 0.0, 0.9, 0.8])

# Note: The cosine function from scipy.spatial.distance calculates the cosine distance,
# which is 1 - cosine similarity. So we will subtract it from 1 to get similarity.
print('cosine(king, queen):', round(1 - cosine(king, queen), 3))   # high
print('cosine(king, apple):', round(1 - cosine(king, apple), 3))   # low
```

```python
import numpy as np
from numpy import linalg as la

A2 = np.array([[3., 2.],
               [1., 4.]])
b2 = np.array([7., 9.])

cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])

# 1. Solve A2 x = b2
x = la.solve(A2, b2)
print("Solution x =", x)

# 2. Rank of A2
rank = la.matrix_rank(A2)
print("Rank of A2 =", rank)

# 3. Cosine similarity
cos_cat_dog = np.dot(cat, dog) / (la.norm(cat) * la.norm(dog))
cos_cat_car = np.dot(cat, car) / (la.norm(cat) * la.norm(car))

print("Cosine(cat, dog) =", cos_cat_dog)
print("Cosine(cat, car) =", cos_cat_car)

if cos_cat_dog > cos_cat_car:
    print("cat is more similar to dog.")
else:
    print("cat is more similar to car.")
```
